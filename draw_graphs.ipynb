```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def draw_graphs(original_result_path, modified_result_path, m0, m):
    origin_=pd.read_csv(original_result_path)
    modified_=pd.read_csv(modified_result_path)

    plt.figure(figsize=(10, 5))

    plt.plot(origin_['recall'], origin_['Cal'], marker='o', color='b', label='Original HNSW')
    plt.plot(modified_['recall'], modified_['Cal'], marker='o', color='r', label='Modified HNSW')


    for x, y, ef in zip(origin_['recall'], origin_['Cal'], origin_['ef']):
        if ef <=5:
            plt.text(x, y, f'ef={ef}', fontsize=10, ha='center', va='bottom')

    for x, y, ef in zip(modified_['recall'], modified_['Cal'], modified_['ef']):
        if ef <=5:
            plt.text(x, y, f'ef={ef}', fontsize=10, ha='center', va='bottom')

    plt.xlabel('Recall')
    plt.ylabel('Number of vertices / avg. calc')
    plt.title(f'Original HNSW vs Modified HNSW(M0={m0}, m={m})')
    plt.legend()
    plt.grid(True)
    plt.show()
```

```python
draw_graphs("./results/original_heuristic_64_32.csv", "./results/modified_heuristic_64_32.csv", 64, 32)
```

```python
draw_graphs("./results/original_heuristic_32_16.csv", "./results/modified_heuristic_32_16.csv", 32, 16)
```
